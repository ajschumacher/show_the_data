```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import pavement
```

```python
singer = pd.read_csv('pavement/singer.csv')
singer
```

```python
pavement.plot(sorted(singer.height), bins=4)
plt.ylim(-2, 2)
plt.show()
```

```python
singer.part.value_counts()
```

```python
bins = 4
pavement.plot(sorted(singer.height[singer.part == 'Soprano 1']), bins=bins, ypos=7)
pavement.plot(sorted(singer.height[singer.part == 'Soprano 2']), bins=bins, ypos=6)
pavement.plot(sorted(singer.height[singer.part == 'Alto 1']), bins=bins, ypos=5)
pavement.plot(sorted(singer.height[singer.part == 'Alto 2']), bins=bins, ypos=4)
pavement.plot(sorted(singer.height[singer.part == 'Tenor 1']), bins=bins, ypos=3)
pavement.plot(sorted(singer.height[singer.part == 'Tenor 2']), bins=bins, ypos=2)
pavement.plot(sorted(singer.height[singer.part == 'Bass 1']), bins=bins, ypos=1)
pavement.plot(sorted(singer.height[singer.part == 'Bass 2']), bins=bins, ypos=0)
```

```python
labels = ['Soprano 1', 'Soprano 2', 'Alto 1', 'Alto 2', 'Tenor 1', 'Tenor 2', 'Bass 1', 'Bass 2']
pavement.multi(singer.height, singer.part, labels=labels, bins=4, show_whiskers=True)
plt.xlabel('Height in inches')
plt.show()
```

```python
pavement.plot(sorted(np.random.randn(1000)), bins=32)
plt.ylim(-2, 2)
plt.show()
```

```python
rng = np.random.default_rng(42)
N = 1000

n  = rng.standard_normal(N)                              # standard normal
s  = rng.lognormal(0, 0.5, N) - 1.0                      # right-skewed (shifted lognormal)
k  = rng.standard_t(df=4, size=N) * 0.5                  # leptokurtic (heavy-tailed t)
mm = np.concatenate([rng.normal(-1, 0.35, N // 2),       # bimodal mixture
                     rng.normal( 1, 0.35, N // 2)])
rng.shuffle(mm)
```

```python
bins=16
pavement.plot(sorted(n), bins=bins, ypos=3)
pavement.plot(sorted(s), bins=bins, ypos=2)
pavement.plot(sorted(k), bins=bins, ypos=1)
pavement.plot(sorted(mm), bins=bins, ypos=0)
plt.gca().set_yticks([0, 1, 2, 3], ['bimodal', 'leptokurtic', 'right-skewed', 'normal'])
plt.show()
```
